```python
import numpy as np
from collections import deque
import heapq, itertools
import random
```

```python
ARRIVAL = 5
SERVICE = 8
users = 0
MAX_TIME = 1000
N_SERVERS = 2
PRIORITY = [0]              # 0 = priorità più alta
PROBS_PRIORITY = [1]  
```

```python
def sample_priority():
    return random.choices(PRIORITY, weights=PROBS_PRIORITY)[0]


def sample_geom(mean):
    return int(np.random.geometric(1.0 / mean))
```

```python
class FutureEventSet():
    
    def __init__(self):
        self.heap = []
        
        
    def insertEvent(self, event):
        heapq.heappush(self.heap, event)
        
        
    def getEvent(self):
        event = heapq.heappop(self.heap)
        return event
```

```python
class Client():
    
    def __init__(self, type: int, arrivalTime: int):
        self.type = type
        self.arrivalTime = arrivalTime
        
        
    def toTuple(self) -> tuple[int, int]:
        return (self.type, self.arrivalTime)
        
```

```python
class UsersQueue:
    def __init__(self, policy: str):
        self.policy = policy
        
        if self.policy == "priority":
            self.queue = []
            
        elif self.policy == "FIFO":
            self.queue = deque()
            
        elif self.policy == "LIFO":
            self.queue = deque()
            
        self.counter = itertools.count()


    def insertUser(self, user: Client) -> None:
        if self.policy == "priority":
            heapq.heappush(self.queue, (user.type, user.arrivalTime, next(self.counter), user)) # type: ignore
            
        elif self.policy == "FIFO":
            self.queue.append(user)
            
        elif self.policy == "LIFO":
            self.queue.append(user)


    def getUser(self) -> Client:
        if self.policy == "priority":
            return heapq.heappop(self.queue)[-1] # type: ignore
        
        elif self.policy == "FIFO":
            return self.queue.popleft() # type: ignore
        
        elif self.policy == "LIFO":
            return self.queue.pop()
            
        else:
            return self.queue.pop()
        
        
    def __len__(self):
        return len(self.queue)
```

```python
def arrival(time: int, FES: FutureEventSet, usersQueue: UsersQueue, counter) -> None:
    global users
    
    arrival_time = time + sample_geom(ARRIVAL)
    FES.insertEvent((arrival_time, "arrival"))
    
    users += 1 # type: ignore
    
    clientType = sample_priority()
    user = Client(clientType, time)
    
    if users <= N_SERVERS:
        service_time = time + sample_geom(SERVICE)
        FES.insertEvent((service_time, "departure"))
        
    else:
        usersQueue.insertUser(user)
        
        
        
def departure(time: int, FES: FutureEventSet, usersQueue: UsersQueue, counter) -> None:
    global users
    
    if len(usersQueue) > 0:
        client = usersQueue.getUser()
    users -= 1 # type: ignore
    
    if users >= N_SERVERS:
        service_time = time + sample_geom(SERVICE)
        FES.insertEvent((service_time, "departure"))
```

```python
users = 0
FES = FutureEventSet()
usersQueue = UsersQueue(policy="LIFO")
time = 0
FES.insertEvent((0, "arrival"))


while FES.heap:
    time, event_type = FES.getEvent()
    
    if time > MAX_TIME:
        break
    
    if event_type == "arrival":
        arrival(time, FES, usersQueue, None)
    
    elif event_type == "departure":
        departure(time, FES,usersQueue, None)
        
    print(f"{event_type} --- Time: {time} --- len(queue): {len(usersQueue)} --- users: {users}\n") # type: ignore
```
